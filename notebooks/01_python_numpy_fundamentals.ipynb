### 1.4 Indexing and slicing

```python
# These allow you to select specific values or parts of a collection.
x = [10, 20, 30, 40]
```

```python
# indexing allows you to access a single value from a collection.
print(x[0])  # first element
print(x[-1]) # last element
```

```python
# slicing allows you to access a range of values from a collection.
# slicing is useful because in dataset we often need to select a subset of data for training and testing.

print(x[1:3])  # elements from index 1 to 2
print(x[:2])   # elements from start to index 1
```

### 1.5 For loops

```python
# for loops are used to iterate over a collection of values.
data = [10, 20, 30, 40]

for value in data:
    print(value)
```

### 1.6 Functions

```python
# A function is a reusable block of code that performs a specific task. It mainly helps to organize code and make it more readable and maintainable.
# Main use cases of functions in machine learning are data preprocessing, model training, and evaluation.

def scale_features(values, scale=1.0):
    """Multiply each value by a scale factor."""
    scaled_values = [value * scale for value in values]
    return scaled_values

features = [0.2, 0.5, 0.9]
scaled_features = scale_features(features, scale=2.0)
print(scaled_features)
```

### 1.7 List Comprehensions

```python
# A list comprehension is a concise way to create a new list by applying an expression to each item in an existing list.
# Here, the order follows, first the for loop, then the if condition, and finally the expression to apply to each value.

# Example: for loop get 128 then check if 128 is between 0 and 255, then apply the function to get 256.
# Example: for loop get -1 then check if -1 is between 0 and 255, then skip the value and do not apply the function.

def some_function(x):
    return x * 2

values = [0, 128, 255, -1, 300]
scaled_values = [some_function(x) for x in values if 0 <= x <= 255]
print(scaled_values)
```
